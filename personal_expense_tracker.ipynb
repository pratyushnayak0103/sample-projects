```python
## import csv
import csv
from datetime import datetime
class ExpenseTracker:
        def __init__(self):
            self.expenses=[] # initialize the list
            self.monthly_budget = 0 # initialize monthly
            self.load_expenses_csv() 
            
        # this function will load the existing csv file 
        def load_expenses_csv(self):            
            try:
                filename="expenses.csv"
                with open(filename, mode='r', newline='', encoding='utf-8') as csvfile:            
                    reader = csv.DictReader(csvfile)
                    for row in reader:                
                        try:
                            row['amount'] = float(row['amount'])
                        except (ValueError, KeyError):                    
                            pass
                        self.expenses.append(row)
                print(f"Successfully loaded expenses from '{filename}'")
            except FileNotFoundError:
                print(color.YELLOW +f"Warning: The file '{filename}' was not found. Please user Option 1 to add expense to get the file created"+color.END)
            except Exception as e:
                print(f"An unexpected error occurred: {e}")
            

        # Function to retrieve and display all stored expenses
        def view_expenses(self):
            for expense in self.expenses:
                print(expense)

        # Function to Set the monthly budget
        def set_monthly_budget(self,user_monthly_budget_input):
            self.monthly_budget = user_monthly_budget_input
            print(f"Monthly expenses has been set to {self.monthly_budget}")

        #  function to add expenses
        def add_expense(self,date, category,amount, description):
            expense = {"date":date,
                      "category":category,
                      "amount":amount,
                      "description":description}
            self.expenses.append(expense)
            print('Expense added')

        # Function to track expenses.    
        def track_expenses(self):
                print(f" Monthly allocated budget - {self.monthly_budget}")                
                expense_current = sum(exp['amount'] for exp in self.expenses)
                print(f" Current expenses incurred - {expense_current}") 
                if expense_current > self.monthly_budget:
                    print('Warning !- Monthly budget exceeded')
                else:
                    print(f"You have {self.monthly_budget-expense_current} left for the month")

        # Function to save expense in a csv file
        def save_expenses_csv(self):
            filename = "expenses.csv"
            if not self.expenses:
                print("No expenses to save.")
                return
        
            # Define the fieldnames (CSV headers) in the desired order
            fieldnames = ['date', 'category', 'amount', 'description']
        
            try:
                with open(filename, 'w', newline='', encoding='utf-8') as csvfile:
                    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
        
                    writer.writeheader()  # Write the header row
                    for expense in self.expenses:                
                        if isinstance(expense.get('date'), datetime):
                            expense['date'] = expense['date'].strftime('%Y-%m-%d')
                        writer.writerow(expense)
                print(f"Expenses successfully saved to '{filename}'")
            except IOError as e:
                print(f"Error saving expenses to CSV: {e}")
class color:
   PURPLE = '\033[95m'
   CYAN = '\033[96m'
   DARKCYAN = '\033[36m'
   BLUE = '\033[94m'
   GREEN = '\033[92m'
   YELLOW = '\033[93m'
   RED = '\033[91m'
   BOLD = '\033[1m'
   UNDERLINE = '\033[4m'
   END = '\033[0m'
    
def main():
    # Initialize the class instance
    tracker = ExpenseTracker()
    while True:
        print()
        print('***********************************************************')
        print('*************** Please select the options  ****************')
        print('***********************************************************')
        print('1,Add Expense')
        print('2,View Expenses')
        print('3,Set monthly budget')
        print('4,Track Expenses')
        print('5,Save Expenses')
        print('6,Exit')
        print('***********************************************************')
        print()
        choice =  input('Please select an option :')
        if choice == '1':
            date = input('Enter date(yyyy-mm-dd):')
            amount = int(input('Enter amount:'))
            category = input('Enter category:')
            description = input('Enter description:')
            tracker.add_expense(date, category, amount, description)
            print()
        elif choice == '2':
            tracker.view_expenses()
        elif choice == '3':
            user_monthly_budget_input = int(input('Enter monthly budget amount:'))
            tracker.set_monthly_budget(user_monthly_budget_input) 
        elif choice == '4':
            tracker.track_expenses()
        elif choice == '5':
            tracker.save_expenses_csv() 
        elif choice == '6':
            print('Exit from program')
            break

if __name__=='__main__':
    main()
```
